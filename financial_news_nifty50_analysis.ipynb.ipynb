```python
def sentiment_score(text):
    text = str(text).lower()
    score = 0
    for word in positive_words:
        score += text.count(word)
    for word in negative_words:
        score -= text.count(word)
    return score
```

```python
# PART B — MACHINE LEARNING PIPELINE

# Create tomorrow's market direction as the prediction target
merged = merged.sort_values('date').reset_index(drop=True)
```

```python
merged['next_day_target'] = merged['target'].shift(-1)

merged[['date', 'target', 'next_day_target']].head(10)
```

```python
merged[['date', 'sentiment_score']].head()
```

```python
merged['sentiment_ma3'] = (
    merged['sentiment_score']
    .rolling(3)
    .mean()
)

merged['sentiment_ma7'] = (
    merged['sentiment_score']
    .rolling(7)
    .mean()
)

merged[['date', 'sentiment_score', 'sentiment_ma3', 'sentiment_ma7']].head(10)
```

```python
merged['sentiment_std3'] = (
    merged['sentiment_score']
    .rolling(3)
    .std()
)

merged[['date', 'sentiment_score', 'sentiment_std3']].head(10)
```

```python
merged['movement_lag1'] = merged['movement'].shift(1)
merged['movement_lag2'] = merged['movement'].shift(2)

merged[['date', 'movement', 'movement_lag1', 'movement_lag2']].head(10)
```

```python
merged['volume_change'] = merged['shares_traded'].pct_change()

merged['high_low_spread'] = merged['high'] - merged['low']

merged[['date', 'volume_change', 'high_low_spread']].head(10)
```

```python
feature_cols = [
    'sentiment_score',
    'sentiment_ma3',
    'sentiment_ma7',
    'sentiment_std3',
    'movement_lag1',
    'movement_lag2',
    'volume_change',
    'high_low_spread'
]

merged_ml = merged.dropna(
    subset=feature_cols + ['next_day_target']
).copy()

print(merged_ml.shape)
```

```python
X = merged_ml[feature_cols]
y = merged_ml['next_day_target']
```

```python
split_idx = int(len(merged_ml) * 0.8)

X_train = X.iloc[:split_idx]
X_test  = X.iloc[split_idx:]

y_train = y.iloc[:split_idx]
y_test  = y.iloc[split_idx:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("Training period:")
print(merged_ml['date'].iloc[:split_idx].min())
print("to")
print(merged_ml['date'].iloc[:split_idx].max())

print("\nTesting period:")
print(merged_ml['date'].iloc[split_idx:].min())
print("to")
print(merged_ml['date'].iloc[split_idx:].max())
```

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=1000)

log_reg.fit(X_train, y_train)

log_pred = log_reg.predict(X_test)

print("Logistic Regression Accuracy:",
      log_reg.score(X_test, y_test))
```

```python
print("Actual:   ", y_test.values[:10])
print("Predicted:", log_pred[:10])
```

```python
from sklearn.metrics import accuracy_score

majority_class = y_train.mode()[0]

baseline_pred = [majority_class] * len(y_test)

baseline_accuracy = accuracy_score(y_test, baseline_pred)

print("Majority class:", majority_class)
print("Baseline Accuracy:", baseline_accuracy)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)

print("Random Forest Accuracy:",
      accuracy_score(y_test, rf_pred))
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    random_state=42,
    eval_metric='logloss'
)

xgb.fit(X_train, y_train)

xgb_pred = xgb.predict(X_test)

print("XGBoost Accuracy:",
      accuracy_score(y_test, xgb_pred))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Classification Report - Random Forest")
print(classification_report(y_test, rf_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, rf_pred))
```

```python
import pandas as pd

feature_importance = pd.DataFrame({
    'Feature': feature_cols,
    'Importance': rf.feature_importances_
}).sort_values('Importance', ascending=False)

print(feature_importance)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.xlabel('Importance')
plt.ylabel('Feature')
plt.title('Random Forest Feature Importance')
plt.gca().invert_yaxis()

plt.show()
```

```python
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV

tscv = TimeSeriesSplit(n_splits=5)

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, None],
    'min_samples_split': [2, 5]
}

rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=tscv,
    scoring='accuracy',
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)

print("Best parameters:", rf_grid.best_params_)
print("Best CV accuracy:", rf_grid.best_score_)
```

```python
best_rf = rf_grid.best_estimator_

best_rf_pred = best_rf.predict(X_test)

best_rf_accuracy = accuracy_score(y_test, best_rf_pred)

print("Tuned Random Forest Accuracy:", best_rf_accuracy)
```

```python
test_results = merged_ml.iloc[split_idx:].copy()

test_results['prediction'] = rf_pred

test_results['strategy_return'] = (
    test_results['movement'] / test_results['open']
) * test_results['prediction']

test_results['market_return'] = (
    test_results['movement'] / test_results['open']
)

test_results['strategy_cumulative'] = (
    1 + test_results['strategy_return']
).cumprod()

test_results['market_cumulative'] = (
    1 + test_results['market_return']
).cumprod()

print(test_results[
    ['date', 'prediction', 'strategy_return',
     'market_return', 'strategy_cumulative',
     'market_cumulative']
].tail())
```

```python
test_results = merged_ml.iloc[split_idx:].copy()

test_results['prediction'] = rf_pred

# Actual return of the same day
test_results['market_return'] = (
    test_results['movement'] / test_results['open']
)

# Tomorrow's return
test_results['next_day_return'] = (
    test_results['market_return'].shift(-1)
)

# Strategy takes tomorrow's return only when today's model predicts UP
test_results['strategy_return'] = (
    test_results['prediction'] * test_results['next_day_return']
)

# Removing the final row because there is no next-day return
test_results = test_results.dropna(
    subset=['next_day_return']
).copy()

test_results['strategy_cumulative'] = (
    1 + test_results['strategy_return']
).cumprod()

test_results['market_cumulative'] = (
    1 + test_results['next_day_return']
).cumprod()

print(test_results[
    ['date', 'prediction', 'next_day_return',
     'strategy_return', 'strategy_cumulative',
     'market_cumulative']
].tail())
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    test_results['date'],
    test_results['strategy_cumulative'],
    label='Random Forest Strategy'
)

plt.plot(
    test_results['date'],
    test_results['market_cumulative'],
    label='Market Buy & Hold'
)

plt.xlabel('Date')
plt.ylabel('Cumulative Value')
plt.title('Random Forest Strategy vs Market Performance')
plt.legend()
plt.grid(True)

plt.show()
```

```python
print("FINAL RESULTS")
print("-" * 40)

strategy_return = test_results['strategy_cumulative'].iloc[-1] - 1
market_return = test_results['market_cumulative'].iloc[-1] - 1

print(f"Strategy Return: {strategy_return:.2%}")
print(f"Market Return:   {market_return:.2%}")

print("\nPrediction Statistics")
print("-" * 40)

correct = (test_results['prediction'] == test_results['next_day_target']).sum()
total = len(test_results)

print("Total predictions:", total)
print("Correct predictions:", correct)
print(f"Prediction Accuracy: {correct/total:.2%}")

print("\nUP predictions:", int(test_results['prediction'].sum()))
print("DOWN predictions:", int((test_results['prediction'] == 0).sum()))
```

# Financial News Sentiment vs NIFTY50 Analysis

## Project Overview

This project investigates whether financial news sentiment and market-related features can help predict the next-day direction of the NIFTY50 index.

The analysis combines financial news sentiment with NIFTY50 market data and evaluates multiple machine learning models for next-day market direction prediction.

# Results and Model Evaluation

Multiple machine learning models were evaluated against a majority-class baseline. The models were assessed using accuracy, classification metrics, confusion matrix analysis, feature importance, and a simple backtesting strategy.

```python
results = pd.DataFrame({
    'Model': [
        'Majority Baseline',
        'Logistic Regression',
        'Random Forest',
        'XGBoost',
        'Tuned Random Forest'
    ],
    'Accuracy': [
        baseline_accuracy,
        log_reg.score(X_test, y_test),
        accuracy_score(y_test, rf_pred),
        accuracy_score(y_test, xgb_pred),
        best_rf_accuracy
    ]
})

results
```

## Model Comparison

Random Forest achieved the highest initial test accuracy of 48%, slightly outperforming the majority-class baseline of 44%. Logistic Regression performed at the same level as the baseline, while XGBoost achieved 40% accuracy.

Time-series cross-validation was used for Random Forest hyperparameter tuning. The tuned model achieved 44% accuracy on the held-out test set, indicating that tuning did not provide a meaningful improvement.

Overall, the models did not demonstrate strong predictive performance for next-day NIFTY50 direction on the available dataset.

# Backtesting

A simple backtesting strategy was used to evaluate whether the model's predictions could translate into improved market performance.

When the model predicted an upward movement, the strategy took the corresponding next-day market return. When the model predicted a downward movement, the strategy remained out of the market.

The strategy was compared with a buy-and-hold market strategy.

```python
backtest_summary = pd.DataFrame({
    'Strategy': [
        'Random Forest Strategy',
        'Market Buy & Hold'
    ],
    'Cumulative Return': [
        test_results['strategy_cumulative'].iloc[-1] - 1,
        test_results['market_cumulative'].iloc[-1] - 1
    ]
})

backtest_summary
```

# Conclusion

The analysis examined whether financial news sentiment combined with NIFTY50 market features could improve next-day market direction prediction.

Among the evaluated models, Random Forest achieved the highest initial test accuracy of 48%, compared with a 44% majority-class baseline. However, the improvement was small, and the tuned Random Forest did not improve the held-out test performance.

Feature importance analysis showed that volume change and high-low spread were among the most important features, while rolling sentiment features such as the 7-day sentiment moving average also contributed to the model.

The backtesting analysis resulted in a strategy return of approximately -2.14%, compared with -1.05% for the market buy-and-hold strategy. Therefore, the model did not outperform the market during the tested period.

Overall, the results suggest that the available financial news sentiment and market features provided only limited predictive information for next-day NIFTY50 direction in this dataset.
